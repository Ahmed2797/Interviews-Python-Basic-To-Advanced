# 📘 Linear Regression – Complete Master Notes

---

## 1. Regression Fundamentals

Regression is a supervised learning technique used to predict a continuous value.

### Equation:
$$
y = mx + b
$$

For multiple variables:
$$
y = b_0 + b_1x_1 + b_2x_2 + \cdots + b_nx_n
$$

---

## 2. Math Foundations

### Mean
$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

### Variance
Measures spread.
$$
\sigma^2 = \frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2
$$

### Standard Deviation
$$
\sigma = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2}
$$

### Covariance

Direction of relationship between variables.

→ same direction
→ opposite direction

$$
cov(x,y) = \frac{1}{n}\sum (x_i - \bar{x})(y_i - \bar{y})
$$

### Correlation
Normalized covariance:
Range:

-1 → perfect negative
0 → no relation
1 → perfect positive

$$
r = \frac{cov(x,y)}{\sigma_x \sigma_y}
$$

Range:
$$
-1 \le r \le 1
$$

### Homoscedasticity

Constant variance of errors.

---

## 3. Simple Linear Regression

### Model:
$$
y = mx + b
$$

### Error:
$$
error = y - \hat{y}
$$

---

## 4. Cost Function

### MAE:
$$
MAE = \frac{1}{n}\sum |y_i - \hat{y}_i|
$$

### MSE:
$$
MSE = \frac{1}{n}\sum (y_i - \hat{y}_i)^2
$$

### RMSE:
$$
RMSE = \sqrt{MSE}
$$

---

## 5. Gradient Descent

### Update Rule:
$$
\theta = \theta - \alpha \frac{\partial J(\theta)}{\partial \theta}
$$

---

## 6. Normal Equation

$$
\theta = (X^TX)^{-1}X^Ty
$$

---

## 7. R² Score

$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$

---

## 8. Adjusted R²

$$
R^2_{adj} = 1 - (1-R^2)\frac{n-1}{n-p-1}
$$

---

## 9. Ridge Regression

$$
Loss = MSE + \alpha \sum w^2
$$

---

## 10. Lasso Regression

$$
Loss = MSE + \alpha \sum |w|
$$

---

## 11. Elastic Net

$$
Loss = MSE + \alpha_1 \sum |w| + \alpha_2 \sum w^2
$$

---

## 12. VIF

$$
VIF = \frac{1}{1 - R^2}
$$

---

## 13. Final Model Equation

$$
\hat{y} = w^T x + b
$$

---

## 14. Gradient Descent Intuition

Minimize:
$$
J(w) = \frac{1}{n}\sum (y - \hat{y})^2
$$

---

## 15. Key Insight

- If α = 0 → Linear Regression
- If α ↑ → model becomes simpler
- If α too high → underfitting


```python
import numpy as np

class LinearRegressionScratch:

    def __init__(self, learning_rate=0.01, epochs=1000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.w = 0.0
        self.b = 0.0
        self.cost_history = []

    def predict(self, X):
        return self.w * X + self.b

    def mse(self, y_true, y_pred):
        return np.mean((y_true - y_pred) ** 2)

    def fit(self, X, y):

        n = len(X)

        for epoch in range(self.epochs):

            predictions = self.predict(X)

            dw = (-2 / n) * np.sum(X * (y - predictions))
            db = (-2 / n) * np.sum(y - predictions)

            self.w -= self.learning_rate * dw
            self.b -= self.learning_rate * db

            cost = self.mse(y, self.predict(X))
            self.cost_history.append(cost)

    def score(self, X, y):

        predictions = self.predict(X)

        ss_total = np.sum((y - np.mean(y)) ** 2)

        ss_res = np.sum((y - predictions) ** 2)

        return 1 - ss_res / ss_total
```

```python
import numpy as np

class LassoRegressionScratch:

    def __init__(self,
                 learning_rate=0.01,
                 epochs=1000,
                 alpha=0.1):

        self.learning_rate = learning_rate
        self.epochs = epochs
        self.alpha = alpha

        self.w = 0.0
        self.b = 0.0

        self.cost_history = []

    def predict(self, X):

        return self.w * X + self.b

    def fit(self, X, y):

        n = len(X)

        for epoch in range(self.epochs):

            predictions = self.predict(X)

            dw = (-2/n) * np.sum(
                X * (y - predictions)
            )

            dw += self.alpha * np.sign(self.w)

            db = (-2/n) * np.sum(
                y - predictions
            )

            self.w -= self.learning_rate * dw
            self.b -= self.learning_rate * db

            mse = np.mean(
                (y - self.predict(X)) ** 2
            )

            penalty = self.alpha * abs(self.w)

            cost = mse + penalty

            self.cost_history.append(cost)

    def score(self, X, y):

        predictions = self.predict(X)

        ss_total = np.sum(
            (y - np.mean(y)) ** 2
        )

        ss_res = np.sum(
            (y - predictions) ** 2
        )

        return 1 - ss_res / ss_total
```

Ordinary Least Squares Linear Regression has very few tunable hyperparameters because it solves for the coefficients analytically. The main parameters are fit_intercept, copy_X, tol, n_jobs, and positive. However, if Linear Regression is implemented using Gradient Descent, then the key hyperparameters become the learning rate, number of epochs, batch size, and initialization. For regularized models like Ridge and Lasso, alpha is the most important hyperparameter because it controls the strength of regularization.
