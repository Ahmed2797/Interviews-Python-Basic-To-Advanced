### What is an Artificial Neural Network (ANN)?

- An Artificial Neural Network (ANN) is a machine learning model inspired by the way the human brain processes information.
It consists of many interconnected artificial neurons, which learn patterns from data by adjusting their weights during training.

- Simple Definition : ANN is a network of connected neurons that learns complex relationships between inputs and outputs.

### What is the role of weights?

- Answer: Weights determine how much influence each input feature has on the neuron's output. During training, the network updates these weights to minimize prediction error.

### Why is bias important?

- Answer: Bias allows the neuron to shift its decision boundary, making the model more flexible and capable of fitting a wider range of patterns.

### What is forward propagation?

- Answer: Forward propagation is the process of passing input data through the network to compute predictions. It involves calculating weighted sums and applying activation functions at each layer.

### What is the difference between a Perceptron and a Multi-Layer Perceptron (MLP)?

- Answer: A Perceptron has a single layer and can solve only linearly separable problems. An MLP has one or more hidden layers and can learn complex non-linear relationships.

### Activation Function Selection Guide

| Problem                    | Output Activation      |
| -------------------------- | ---------------------- |
| Regression                 | Linear (no activation) |
| Binary Classification      | Sigmoid                |
| Multi-class Classification | Softmax                |
| Hidden Layers              | ReLU (default)         |

### Comparison Table
| Activation | Range            | Best Use            | Main Issue                            |
| ---------- | ---------------- | ------------------- | ------------------------------------- |
| Sigmoid    | 0 to 1           | Binary output       | Vanishing gradient                    |
| Tanh       | -1 to 1          | Older hidden layers | Vanishing gradient                    |
| ReLU       | 0 to ∞           | Hidden layers       | Dying ReLU                            |
| Leaky ReLU | -∞ to ∞          | Hidden layers       | Extra hyperparameter (negative slope) |
| ELU        | -α to ∞          | Hidden layers       | Slower computation                    |
| Softmax    | 0 to 1 (sum = 1) | Multi-class output  | Can become overconfident              |

# Cheat Sheet (Must Memorize)

| Activation     | Formula                                                                     | Range            | Typical Use                       |
| -------------- | --------------------------------------------------------------------------- | ---------------- | --------------------------------- |
| **Sigmoid**    | $\displaystyle \sigma(z)=\frac{1}{1+e^{-z}}$                                | 0 to 1           | Binary classification output      |
| **Tanh**       | $\displaystyle \tanh(z)=\frac{e^z-e^{-z}}{e^z+e^{-z}}$                      | -1 to 1          | Older hidden layers               |
| **ReLU**       | $\displaystyle \text{ReLU}(z)=\max(0,z)$                                    | 0 to ∞           | Hidden layers                     |
| **Leaky ReLU** | $\displaystyle \begin{cases} z, & z>0 \\ 0.01z, & z\le0 \end{cases}$         | -∞ to ∞          | Hidden layers                     |
| **ELU**        | $\displaystyle \begin{cases} z, & z>0 \\ \alpha(e^z-1), & z\le0 \end{cases}$ | -α to ∞          | Hidden layers                     |
| **Softmax**    | $\displaystyle P_i=\frac{e^{z_i}}{\sum_j e^{z_j}}$                          | 0 to 1 (sum = 1) | Multi-class classification output |

## Activation Function Selection Guide

* **Hidden layers:** ReLU is the default choice; Leaky ReLU or ELU are good alternatives if the Dying ReLU problem is an issue.
* **Binary classification output:** Sigmoid.
* **Multi-class classification output:** Softmax.
* **Regression output:** Linear (no activation).

### Q1. Why do we need activation functions?

**Answer (English):**
- Activation functions introduce non-linearity, allowing neural networks to learn complex relationships. Without them, multiple linear layers collapse into a single linear transformation.

**উত্তর (বাংলা):**
- Activation Function নিউরাল নেটওয়ার্কে **Non-linearity (অরৈখিকতা)** যোগ করে, যার ফলে এটি জটিল সম্পর্ক (Complex Relationships) শিখতে পারে। যদি Activation Function না থাকে, তাহলে যতগুলো Linear Layer-ই ব্যবহার করা হোক না কেন, সেগুলো শেষ পর্যন্ত একটি মাত্র Linear Transformation-এ পরিণত হবে।

### Q2. Why isn't a linear activation enough?

**Answer (English):**
- A composition of linear functions is still linear, so the network cannot model nonlinear patterns regardless of its depth.

**উত্তর (বাংলা):**
- একাধিক Linear Function একসাথে ব্যবহার করলেও ফলাফল Linear-ই থাকে। তাই নেটওয়ার্ক যত গভীর (Deep) হোক না কেন, এটি Non-linear Pattern শিখতে পারে না।

### Q3. Why is ReLU the most popular hidden-layer activation?

**Answer (English):**
- ReLU is computationally simple, reduces the vanishing gradient problem, and enables efficient training of deep neural networks.

**উত্তর (বাংলা):**
- ReLU সবচেয়ে জনপ্রিয় কারণ এটি:
- গণনা (Computation) করা খুব সহজ।
- Vanishing Gradient Problem অনেকটাই কমায়।
- Deep Neural Network দ্রুত এবং কার্যকরভাবে Train করতে সাহায্য করে।

### Q4. What is the Dying ReLU problem?

**Answer (English):**
- If a ReLU neuron consistently receives negative inputs, it outputs zero and may stop updating because its gradient becomes zero.

**উত্তর (বাংলা):**
- যদি একটি ReLU নিউরন বারবার Negative Input পায়, তাহলে তার Output সবসময় **0** হবে। ফলে Gradient-ও **0** হয়ে যায় এবং সেই নিউরন আর শেখা (Learning) বা Weight Update করা বন্ধ করে দেয়। একে **Dying ReLU Problem** বলা হয়।

---

### Q5. When should you use Sigmoid?

**Answer (English):**

**উত্তর (বাংলা):**

**উত্তর (বাংলা):**

### Q1. What is a Loss Function?

Answer:
A loss function measures the difference between the actual and predicted values, providing a numerical value that the model minimizes during training.

### Q2. What is the difference between Loss Function and Cost Function?

Answer:
A loss function measures error for a single training example, while a cost function is the average loss across the entire dataset.

### Q3. Why is MSE sensitive to outliers?

Answer:
Because it squares the errors, larger errors contribute much more to the loss than smaller ones.

### Q4. Why is MAE more robust to outliers?

Answer:
MAE uses the absolute value of errors instead of squaring them, so large errors are not disproportionately amplified.

### Q5. Why is Binary Cross Entropy preferred over MSE for binary classification?

Answer:
Binary Cross Entropy provides stronger gradients for incorrect predictions and aligns well with probabilistic outputs from the Sigmoid activation.

### Q6. What is the difference between Categorical Cross Entropy and Sparse Categorical Cross Entropy?

Answer:
Categorical Cross Entropy requires one-hot encoded labels, while Sparse Categorical Cross Entropy works directly with integer class labels.

````python
import numpy as np

# Input
x = 2

# Actual Output
y = 10

# Initial Weight
w = 3

# Bias
b = 1

# Learning Rate
lr = 0.1

# Forward Pass
y_pred = w * x + b

# Loss
loss = (y - y_pred) ** 2

# Gradient
dw = 2 * (y_pred - y) * x

# Weight Update
w = w - lr * dw

print("Prediction:", y_pred)
print("Loss:", loss)
print("Updated Weight:", w)
````

## Q1. What is Backpropagation?

### Answer (English)

Backpropagation is an algorithm that computes the gradients of the loss function with respect to every weight and bias in a neural network so they can be updated to minimize the loss.

### উত্তর (বাংলা)

**Backpropagation** হলো একটি অ্যালগরিদম যা নিউরাল নেটওয়ার্কের প্রতিটি **Weight** এবং **Bias**-এর জন্য Loss Function-এর **Gradient** গণনা করে। এরপর সেই Gradient ব্যবহার করে Weight ও Bias আপডেট করা হয়, যাতে Loss ধীরে ধীরে কমে যায় এবং Model আরও ভালো Prediction করতে পারে।

> **সহজ ভাষায়:** Backpropagation প্রতিটি Weight-কে বলে দেয়, "তুমি কতটা ভুলের জন্য দায়ী এবং কতটা পরিবর্তন হওয়া উচিত।"

## Q2. What is a Gradient?

### Answer (English)

A gradient measures how much the loss changes with respect to a small change in a model parameter. It indicates the direction and magnitude for updating the parameter.

### উত্তর (বাংলা)

**Gradient** হলো এমন একটি মান, যা বলে দেয় **Weight বা Bias সামান্য পরিবর্তন করলে Loss কতটা পরিবর্তিত হবে।**

Gradient আমাদের দুটি তথ্য দেয়—

- কোন দিকে Weight পরিবর্তন করতে হবে।
- কতটুকু পরিবর্তন করতে হবে।

### Formula

```text
Gradient = ∂L / ∂w
```

যেখানে,

- **L = Loss**
- **w = Weight**

---

## Q3. What is Gradient Descent?

### Answer (English)

Gradient Descent is an optimization algorithm that updates model parameters by moving them in the direction opposite to the gradient to minimize the loss function.

### উত্তর (বাংলা)

**Gradient Descent** হলো একটি Optimization Algorithm, যা Gradient ব্যবহার করে Weight Update করে এবং Loss কমানোর চেষ্টা করে।

এটি সবসময় Gradient-এর বিপরীত দিকে Weight পরিবর্তন করে।

### Formula

```text
w_new = w_old - η × (∂L/∂w)
```

যেখানে,

- **η (Eta)** = Learning Rate
- **∂L/∂w** = Gradient

---

## Q4. Why is the Chain Rule important in Backpropagation?

### Answer (English)

The Chain Rule allows gradients to be propagated backward through multiple layers, enabling the computation of gradients for every parameter in the network.

### উত্তর (বাংলা)

নিউরাল নেটওয়ার্কে অনেকগুলো Layer থাকে।

**Chain Rule** ব্যবহার করে Output Layer থেকে Error ধাপে ধাপে পিছনের Layer-এ পাঠানো হয় এবং প্রতিটি Weight-এর Gradient বের করা হয়।

Chain Rule না থাকলে Deep Neural Network-এর Gradient বের করা সম্ভব হতো না।

### Formula

```text
dy/dx = (dy/dg) × (dg/dx)
```

## Q5. What happens if the learning rate is too high?

### Answer (English)

The optimization may overshoot the minimum, causing unstable training or divergence.

### উত্তর (বাংলা)

যদি **Learning Rate অনেক বেশি হয়**, তাহলে Model একবারে অনেক বড় Step নেয়।

ফলে—

- Minimum Point অতিক্রম করে যায় (Overshoot)
- Loss কমার বদলে বাড়তে পারে
- Training Unstable হয়ে যায়
- কখনও কখনও Model একেবারেই Converge করে না

## Q6. What happens if the learning rate is too low?

### Answer (English)

Training becomes very slow and may require many epochs to converge.

### উত্তর (বাংলা)

যদি **Learning Rate খুব কম হয়**, তাহলে প্রতিবার Weight খুব অল্প পরিবর্তিত হয়।

ফলে—

- Training খুব ধীরে হয়
- অনেক বেশি Epoch লাগে
- Model শেখার গতি কমে যায়

# Backpropagation vs Gradient Descent

| Backpropagation | Gradient Descent |
|-----------------|------------------|
| Computes gradients | Updates weights |
| Uses Chain Rule | Uses Learning Rate |
| Calculates `∂L/∂w` | Applies `w = w - η × (∂L/∂w)` |

### বাংলা ব্যাখ্যা

| Backpropagation | Gradient Descent |
|-----------------|------------------|
| Gradient গণনা করে | Weight Update করে |
| Chain Rule ব্যবহার করে | Learning Rate ব্যবহার করে |
| `∂L/∂w` বের করে | `w = w - η × (∂L/∂w)` সূত্র ব্যবহার করে নতুন Weight বের করে |

# Interview Tips (মনে রাখার শর্টকাট)

An optimizer is an algorithm that updates a model's parameters (weights and biases) using gradients to minimize the loss function.

**Optimizer** হলো একটি অ্যালগরিদম, যা **Gradient** ব্যবহার করে Model-এর **Weight** এবং **Bias** আপডেট করে, যাতে **Loss Function** ধীরে ধীরে কমে এবং Model আরও ভালো Prediction করতে পারে।

## Q2. What is the difference between Batch Gradient Descent and SGD?

### Answer (English)

Batch Gradient Descent updates weights after processing the entire dataset, while SGD updates weights after each individual training sample.

### উত্তর (বাংলা)

**Batch Gradient Descent** পুরো Dataset একবার প্রসেস করার পর একবার Weight Update করে।

অন্যদিকে, **Stochastic Gradient Descent (SGD)** প্রতিটি Training Sample প্রসেস করার পরই Weight Update করে।

### Example

Dataset:

```text
1000 Samples
```

Batch Gradient Descent

```text
1000 Samples → 1 Weight Update
```

SGD

```text
1000 Samples → 1000 Weight Updates
```

## Q4. What is Momentum?

### Answer (English)

Momentum accumulates previous gradients to smooth updates, reduce oscillations, and speed up convergence.

### উত্তর (বাংলা)

**Momentum** পূর্বের Gradient-এর তথ্য (Velocity) মনে রাখে এবং সেই অনুযায়ী Weight Update করে।

এর ফলে—

- Zig-Zag Movement কমে যায়।
- Training দ্রুত হয়।
- দ্রুত Minimum Point-এ পৌঁছানো যায়।

### Formula

```text
v = β × v - η × gradient

w = w + v
```

যেখানে,

- **v** = Velocity
- **β** = Momentum Coefficient
- **η** = Learning Rate

## Q6. Why is RMSProp better than AdaGrad?

### Answer (English)

RMSProp prevents the learning rate from shrinking continuously by using an exponentially weighted average of squared gradients instead of accumulating all past gradients.

### উত্তর (বাংলা)

AdaGrad-এর একটি সমস্যা হলো Learning Rate ধীরে ধীরে খুব ছোট হয়ে যায়।

**RMSProp** এই সমস্যা সমাধান করে।

এটি সব পুরনো Gradient জমা না রেখে **Moving Average** ব্যবহার করে।

ফলে—

- Learning Rate খুব ছোট হয়ে যায় না।
- Training দ্রুত হয়।
- Stable Convergence পাওয়া যায়।

---

## Q7. Why is Adam so popular?

### Answer (English)

Adam combines Momentum and RMSProp, providing adaptive learning rates, fast convergence, and strong performance across many deep learning applications.

### উত্তর (বাংলা)

**Adam (Adaptive Moment Estimation)** সবচেয়ে জনপ্রিয় Optimizer কারণ এটি—

- Momentum ব্যবহার করে।
- RMSProp ব্যবহার করে।
- Adaptive Learning Rate ব্যবহার করে।
- দ্রুত Converge করে।
- অধিকাংশ Deep Learning সমস্যায় খুব ভালো কাজ করে।

তাই নতুন Project শুরু করলে সাধারণত **Adam**-ই Default Choice।

---

## Q8. What is the difference between Adam and AdamW?

### Answer (English)

AdamW decouples weight decay from the gradient update, resulting in better regularization and improved performance, especially for Transformer models.

### উত্তর (বাংলা)

**AdamW** হলো Adam-এর উন্নত সংস্করণ।

মূল পার্থক্য হলো—

- **Adam** Weight Decay-কে Gradient Update-এর সাথে একসাথে পরিচালনা করে।
- **AdamW** Weight Decay-কে আলাদা (Decoupled) করে পরিচালনা করে।

ফলে—

- Regularization আরও ভালো হয়।
- Overfitting কম হয়।
- Transformer Model (যেমন BERT, GPT)-এ AdamW বেশি ব্যবহৃত হয়।

## Regularization & Generalization

---

## Q1. What is Overfitting?

### Answer (English)

Overfitting occurs when a model learns the training data, including noise, too well, resulting in poor performance on unseen data.

### উত্তর (বাংলা)

**Overfitting** হলো এমন একটি অবস্থা যেখানে Model Training Data-এর আসল Pattern-এর পাশাপাশি Noise-ও মুখস্থ করে ফেলে। ফলে Training Accuracy অনেক বেশি হলেও নতুন (Unseen/Test) Data-তে ভালো ফলাফল দিতে পারে না।

### Example

```text
Training Accuracy = 99%
Testing Accuracy  = 72%
```

**Interview Trick**

- High Training Accuracy
- Low Testing Accuracy
- Poor Generalization

---

## Q2. What is Underfitting?

### Answer (English)

Underfitting occurs when a model is too simple to capture the underlying patterns in the data, leading to poor performance on both training and testing datasets.

### উত্তর (বাংলা)

**Underfitting** তখন ঘটে যখন Model এতটাই সহজ (Simple) হয় যে এটি Training Data-এর Pattern-ই শিখতে পারে না।

ফলে Training এবং Testing—দুই ক্ষেত্রেই খারাপ Performance দেয়।

### Example

```text
Training Accuracy = 65%
Testing Accuracy  = 62%
```

**Interview Trick**

- Low Training Accuracy
- Low Testing Accuracy
- Model is too simple

---

## Q3. What is Generalization?

### Answer (English)

Generalization is a model's ability to perform well on new, unseen data after training.

### উত্তর (বাংলা)

**Generalization** হলো একটি Model-এর এমন ক্ষমতা, যার মাধ্যমে এটি Training Data ছাড়াও নতুন (Unseen/Test) Data-তে ভালো Prediction করতে পারে।

### Example

```text
Training Accuracy = 97%
Testing Accuracy  = 95%
```

**Interview Trick**

Good Model = Good Generalization

---

## Q4. What is the difference between Bias and Variance?

### Answer (English)

Bias is the error caused by overly simple assumptions (underfitting), while variance is the error caused by excessive sensitivity to the training data (overfitting).

### উত্তর (বাংলা)

**Bias** হলো এমন Error, যা Model খুব বেশি Simple হওয়ার কারণে হয়।

**Variance** হলো এমন Error, যা Model Training Data-এর উপর অতিরিক্ত নির্ভরশীল হওয়ার কারণে হয়।

| Bias | Variance |
|------|----------|
| Underfitting | Overfitting |
| Simple Model | Complex Model |
| Misses Patterns | Learns Noise |

---

## Q5. What is Dropout?

### Answer (English)

Dropout is a regularization technique that randomly disables a fraction of neurons during training to reduce overfitting and improve generalization.

### উত্তর (বাংলা)

**Dropout** হলো একটি Regularization Technique যেখানে Training-এর সময় কিছু Neuron-কে এলোমেলোভাবে (Randomly) বন্ধ করে দেওয়া হয়।

এর ফলে—

- Overfitting কমে।
- Neuron একে অপরের উপর অতিরিক্ত নির্ভরশীল হয় না।
- Model ভালোভাবে Generalize করতে শেখে।

### Example

```text
Before Dropout

● ● ● ● ●

After Dropout (40%)

● X ● X ●
```

---

## Q6. What is Early Stopping?

### Answer (English)

Early Stopping is a technique that stops training when the validation performance stops improving, preventing overfitting.

### উত্তর (বাংলা)

**Early Stopping** এমন একটি Technique যেখানে Validation Loss বা Validation Accuracy আর উন্নতি না করলে Training বন্ধ করে দেওয়া হয়।

এটি Overfitting প্রতিরোধ করে।

### Example

```text
Epoch 5  → Val Loss = 0.30

Epoch 10 → Val Loss = 0.18 ✅

Epoch 20 → Val Loss = 0.26 ❌
```

Training should stop around **Epoch 10**.

## Q7. What is Batch Normalization?

### Answer (English)

Batch Normalization normalizes the inputs of each layer during training, improving training stability and convergence.

### উত্তর (বাংলা)

**Batch Normalization** প্রতিটি Layer-এর Output Normalize করে।

এর ফলে—

- Training দ্রুত হয়।
- Learning Stable হয়।
- Higher Learning Rate ব্যবহার করা যায়।
- অনেক ক্ষেত্রে Overfitting কিছুটা কমে।

## Q8. What is the difference between L1 and L2 Regularization?

### Answer (English)

| L1 Regularization | L2 Regularization |
|-------------------|-------------------|
| Uses absolute values | Uses squared values |
| Can set weights to zero | Shrinks weights toward zero |
| Performs feature selection | Improves stability |

### উত্তর (বাংলা)

| L1 Regularization | L2 Regularization |
|-------------------|-------------------|
| Absolute Value ব্যবহার করে | Squared Value ব্যবহার করে |
| কিছু Weight সম্পূর্ণ 0 করে দেয় | Weight ছোট করে কিন্তু 0 করে না |
| Feature Selection করে | Model-কে Stable রাখে |

## Q9. Why is Data Augmentation useful?

### Answer (English)

Data Augmentation creates new training examples by transforming existing data, helping reduce overfitting and improving model robustness.

### উত্তর (বাংলা)

**Data Augmentation** মূল Data পরিবর্তন (Rotate, Flip, Zoom, Crop ইত্যাদি) করে নতুন Training Sample তৈরি করে।

এর ফলে—

- Dataset বড় হয়।
- Overfitting কমে।
- Model আরও Robust হয়।

### Common Image Augmentations

- Rotate
- Flip
- Zoom
- Crop
- Shift
- Brightness
