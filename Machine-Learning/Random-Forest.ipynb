## Random Forest
- Random: Each tree uses random samples of the data and random subsets of features.
- Forest: It is a collection of many Decision Trees.
- Random Forest uses Bagging (Bootstrap Aggregating), where multiple Decision Trees are trained on different bootstrap samples, and their predictions are combined using majority voting (classification) or averaging (regression).

```bash
Random Forest is an ensemble learning algorithm based on bagging. It creates multiple bootstrap samples, trains one Decision Tree on each sample, randomly selects features at every split, and combines predictions using majority voting for classification or averaging for regression. This reduces overfitting and improves generalization.
```

### Bagging = Bootstrap Aggregating

- Bootstrap → Create many random datasets from the original dataset.
- Aggregating → Combine the predictions from all trees.

- Bagging = Bootstrap + Aggregation

### What is Bagging?
- Bagging (Bootstrap Aggregating) trains multiple models on different bootstrap samples and combines their predictions to improve accuracy and reduce variance.

### What is Bootstrap Sampling?
- Answer: Bootstrap sampling is the process of creating multiple datasets by randomly sampling the original dataset with replacement.
- After selecting a row, it is put back into the dataset, so it can be selected again. Therefore, some rows may appear multiple times, while others may not appear at all.

| Randomness               | Purpose                                     |
| ------------------------ | ------------------------------------------- |
| Bootstrap Sampling       | Different training data for each tree       |
| Random Feature Selection | Different features considered at each split |

### What happens if max_features is too high?
- Answer: Trees become more similar, increasing correlation and the risk of overfitting.

### What happens if max_features is too low?
- Answer: Trees become more diverse and overfitting decreases, but each tree may be weaker, which can reduce overall accuracy if set too low.

### Advantages
    High accuracy
    Reduces overfitting
    Handles large datasets
    Works for classification and regression
    Handles nonlinear relationships well
    Gives feature importance
    
### Disadvantages
    Slower than a single Decision Tree
    Uses more memory
    Less interpretable ("black box")
    Training time increases with many trees

### Math Split Q/A

- Gini Impurity (Classification)
- Lower Gini means a purer node.
- lower entropy = purer node.
- Higher Information Gain = Better split.

### Out-of-Bag (OOB) Error ⭐ (Very Important)

During bootstrap sampling, not every row is selected.

On average:

- About 63.2% of the original samples appear in one bootstrap dataset.
About 36.8% are left out.

Those left-out samples are called Out-of-Bag (OOB) samples.

- Random Forest can use these OOB samples to estimate model performance without creating a separate validation set.

### What is an OOB sample?
- Answer: An Out-of-Bag sample is a training example that was not selected in a particular bootstrap sample. It can be used to estimate model performance without a separate validation dataset.

### Difference Between min_samples_split and min_samples_leaf

| Parameter           | Meaning                                                           |
| ------------------- | ----------------------------------------------------------------- |
| `min_samples_split` | Minimum samples needed **before a node can split**                |
| `min_samples_leaf`  | Minimum samples that **must remain in each leaf after the split** |

### Hyperparameter Summary

| Hyperparameter      | Increase                        | Decrease                       |
| ------------------- | ------------------------------- | ------------------------------ |
| `n_estimators`      | More stable, slower             | Faster, less stable            |
| `max_depth`         | More complex, overfit risk      | Simpler, underfit risk         |
| `max_features`      | Stronger but more similar trees | More random, lower correlation |
| `min_samples_split` | Simpler tree                    | More complex tree              |
| `min_samples_leaf`  | Smoother predictions            | More flexible, overfit risk    |
| `bootstrap`         | More diversity                  | Less diversity                 |
| `criterion`         | Changes split strategy          | Depends on task                |
| `oob_score`         | Internal validation             | Disabled                       |
| `class_weight`      | Helps imbalance                 | Default equal weights          |
| `random_state`      | Reproducible results            | Different results each run     |

| Parameter           | মনে রাখার সহজ উপায়                                      |
| ------------------- | -------------------------------------------------------- |
| `n_estimators`      | 🌳 **কতগুলো Tree হবে**                                   |
| `max_depth`         | 🌲 **Tree কত গভীর হবে**                                  |
| `max_features`      | 📋 **প্রতি split-এ কত Feature দেখবে**                    |
| `min_samples_split` | ✂️ **Split করতে কমপক্ষে কত Sample লাগবে**                |
| `min_samples_leaf`  | 🍃 **শেষ Leaf-এ কমপক্ষে কত Sample থাকবে**                |
| `bootstrap`         | 🎲 **Random data sampling হবে কি না**                    |
| `criterion`         | 🧮 **Gini না Entropy দিয়ে split করবে**                  |
| `oob_score`         | 📊 **Extra validation ছাড়াই performance estimate**      |
| `class_weight`      | ⚖️ **Imbalanced data-তে minority class-কে গুরুত্ব দেয়** |
| `random_state`      | 🔁 **একই result বারবার পাওয়ার জন্য**                    |


### What is the difference between min_samples_split and min_samples_leaf?

- Answer: min_samples_split controls whether a node can split, while min_samples_leaf controls the minimum number of samples that each resulting leaf must contain.
