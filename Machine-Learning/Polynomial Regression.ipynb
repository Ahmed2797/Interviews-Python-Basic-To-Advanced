## Introduction to Polynomial Regression

### What is Polynomial Regression?

- Polynomial Regression is an extension of Linear Regression that models non-linear relationships by adding polynomial terms (such as x^2,x^3,x^4) while remaining linear in the model parameters.

- Polynomial Regression হলো Linear Regression-এর একটি extension,ইত্যাদি feature যোগ করে non-linear data fit করা হয়।

| Linear Regression        | Polynomial Regression         |
| ------------------------ | ----------------------------- |
| Straight line            | Curved line                   |
| Linear relationship      | Non-linear relationship       |
| Features: (x)            | Features: (x, x^2, x^3,\dots) |
| Simpler                  | More flexible                 |
| May underfit curved data | Can fit curves well           |

### Q1. Is Polynomial Regression a Linear Regression algorithm?

English:
- Yes. Although the relationship between the input and output is non-linear, the model is still linear with respect to its coefficients.

বাংলা:
হ্যাঁ। যদিও input ও output-এর সম্পর্ক curve আকারের হতে পারে, model-এর coefficients-এর ক্ষেত্রে equation এখনো linear। তাই এটিকে Linear Regression পরিবারের অংশ ধরা হয়।

### How is Polynomial Regression different from Linear Regression?

English:
Linear Regression uses only the original features. Polynomial Regression creates additional polynomial features and then fits a Linear Regression model on those transformed features.

বাংলা:
Linear Regression শুধু original feature ব্যবহার করে। Polynomial Regression আগে নতুন feature তৈরি করে, তারপর সেই feature-এর উপর Linear Regression চালায়।

| Hyperparameter        | Importance | Effect                           |
| --------------------- | ---------- | -------------------------------- |
| `degree`              | ⭐⭐⭐⭐⭐      | Controls model complexity        |
| `interaction_only`    | ⭐⭐⭐        | Uses only interaction terms      |
| `include_bias`        | ⭐⭐⭐        | Adds/removes constant feature    |
| `order`               | ⭐          | Memory layout only               |
| `alpha` (Ridge/Lasso) | ⭐⭐⭐⭐⭐      | Controls regularization strength |
| `fit_intercept`       | ⭐⭐⭐⭐       | Learns intercept term            |
| `positive`            | ⭐⭐         | Forces non-negative coefficients |

### What is Degree of Freedom?

- It is the number of independent values that are free to vary after applying constraints.

- Degree of Freedom is the number of independent pieces of information available after accounting for the estimated parameters or constraints. In Linear Regression, the residual degrees of freedom are n−p−1, where n is the number of observations and p is the number of features.
