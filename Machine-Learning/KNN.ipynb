## KNN

### What is K-Nearest Neighbors (KNN)?

- K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for: Classification and Regression both problem statement.
- "Similar things are usually close to each other."
- KNN predicts the output by looking at the nearest data points.

New data
    |
    ↓
Calculate distance(No training time & it stores data.)
    |
    ↓
Find neighbors
    |
    ↓
Predict

- KNN = Lazy Learner

| Property                | KNN            |
| ----------------------- | -------------- |
| Learning Type           | Supervised     |
| Training Style          | Lazy           |
| Model Type              | Non-parametric |
| Classification          | Yes            |
| Regression              | Yes            |
| Feature Scaling Needed  | Yes            |
| Handles Non-linear Data | Yes            |

### Is KNN a parametric algorithm?
Answer:
- No. KNN is a non-parametric algorithm because it does not assume any distribution.

### Q2. Why KNN is lazy learner?

Answer:
- Because it does not learn during training. It stores data and performs computation during prediction.

### Q3. Does KNN require scaling?

Answer:
- Yes, because it uses distance calculations.

Q1. What happens if K is too small?

Answer:

- Small K causes overfitting because the model becomes sensitive to noise.

Q2. What happens if K is too large?

Answer:

- Large K causes underfitting because the model becomes too simple.

Q3. Why is odd K preferred?

Answer:

- To avoid tie situations in binary classification.

| Parameter   | Purpose             | Common Values                   |

| metric      | Distance method     | euclidean, manhattan, minkowski |

### Difference between uniform and distance weighting?

Answer:
- Uniform gives equal importance to all neighbors. Distance weighting gives more importance to closer neighbors.

    K → How many friends?
    weights → Who gets more vote?
    metric → How measure friendship?
    p → Distance style
    algorithm → How find friends faster?
    leaf_size → Tree speed control
