## Support Vector Machine (SVM)?

### What is Support Vector Machine (SVM)?

- A Support Vector Machine (SVM) is a supervised machine learning algorithm used for:
1. Classification ✅ ### It is mainly used for classification.
2. Regression (SVR) ✅
3. Outlier Detection ✅
- Its main goal is to find the best boundary (hyperplane) that separates different classes with the largest possible margin.

- SVM is a supervised machine learning algorithm that finds the optimal separating hyperplane by maximizing the margin between classes.

1. It finds the best separator.
2. It maximizes the margin.
3. It often generalizes better on unseen data.

This is especially useful when classes are clearly separable.

- The Margin is the distance between the boundary and the nearest data points from both classes.
- Generalization means: Performing well on unseen data.

### Why is a larger margin better?
- Because it makes the model more robust to noise and small variations in the data.

### Hyperplane (The Decision Boundary of SVM) ⭐⭐⭐⭐⭐

- Hyperplane is the foundation of SVM.
- SVM finds the best hyperplane that separates classes with maximum margin.

### Logistic Regression vs SVM Hyperplane
- Logistic Regression : Boundary that minimizes probability error
- SVM : Boundary that maximizes margin

✅ Support Vectors are the closest points to the hyperplane
✅ They determine the position of the decision boundary
✅ Only Support Vectors affect the SVM model
✅ Removing normal points usually does not change the model
✅ Removing Support Vectors can change the boundary
✅ They lie on the margin boundaries wTx+b=±1

### Hard Margin SVM ⭐⭐⭐⭐⭐

- Hard Margin SVM is the original idea of SVM.It works when data can be perfectly separated without any mistakes.
- Hard Margin SVM finds a hyperplane that perfectly separates two classes with zero classification errors while maximizing the margin.

- - Perfect Separation + Maximum Margin + Zero Error

✅ Hard Margin is sensitive to noise and outliers
✅ Soft Margin solves this limitation

- Soft Margin SVM is the practical version of SVM.
- Hard Margin says: "No mistakes allowed."
- Soft Margin says: "Allow some mistakes, but keep the boundary strong."

1. Slack variable measures how much a point violates the margin.

### What is C Parameter?
- C is the most important SVM hyperparameter.
- How much do we punish mistakes?
- C is a regularization parameter in SVM that controls how much penalty is given to misclassified samples


| C Value | Margin | Error | Bias   | Variance |
| ------- | ------ | ----- | ------ | -------- |
| Low C   | Large  | More  | Higher | Lower    |
| High C  | Small  | Less  | Lower  | Higher   |

| C        | Margin   | Training Error | Bias   | Variance |
| -------- | -------- | -------------- | ------ | -------- |
| Low C    | Large    | High           | High   | Low      |
| Medium C | Balanced | Medium         | Medium | Medium   |
| High C   | Small    | Low            | Low    | High     |


### What is Slack Variable?
- Slack variable (ξ) measures the amount of margin violation or misclassification allowed for a data point.

### What does C control in SVM?
- C controls the tradeoff between maximizing the margin and minimizing classification errors.

### What happens when C increases?
- The model penalizes errors more, creates a smaller margin, and may overfit.

### What happens when C decreases?
- The model allows more errors, creates a larger margin, and usually generalizes better.

### If C is extremely high, what happens?

- The model tries to classify almost every training sample correctly, producing a narrow margin and increasing the risk of overfitting.

### If training accuracy is 99% but validation accuracy is 75%, what will you do with C?

- The model is overfitting. Reduce C to increase regularization and create a larger margin.
