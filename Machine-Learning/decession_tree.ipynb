## Decision Tree
- A Decision Tree is a supervised machine learning algorithm that makes decisions by asking a series of questions by recursively splitting data into smaller groups based on feature values. It looks like an upside-down tree.

### Root Node 🌱
1. The first question.
2. Represents the entire dataset.
3. There is only one root node.

### Parent Node 👨‍👧
- A node that splits into other nodes.

### Child Node 👶
- A node created from a parent node.

### Leaf Node 🍃
- The final prediction.

### Branch 🌿
- The connection between nodes.

### Q1. What is a leaf node?
- Answer: A node with no children that contains the final prediction.

### Q2. What is a pure node?
- Answer: A node where all samples belong to the same class.

### Q3. How does a Decision Tree decide where to split?
- Answer: It evaluates many possible splits and chooses the one that best separates the data using measures such as Gini Index or Information Gain.

### Q4. Can a Decision Tree have multiple root nodes?
- Answer: No. A Decision Tree always has exactly one root node.

### 🌳 Decision Tree - Entropy, Information Gain & Gini Index
- This is the heart of Decision Trees.

- Entropy → ডেটা কতটা মিশ্রিত বা অনিশ্চিত তা মাপে।
- Information Gain → Split করার পরে Entropy কতটা কমেছে তা মাপে।
- Gini Index → Entropy-এর মতোই Impurity Measure, তবে এটি দ্রুত গণনা করা যায়।
- Decision Tree-এর লক্ষ্য → এমন Split নির্বাচন করা যাতে Child Nodes যতটা সম্ভব Pure হয়। এই কারণেই Decision Tree Information Gain সর্বোচ্চ বা Gini Impurity সর্বনিম্ন এমন Split বেছে নেয়।

### Entropy
- Entropy measures disorder or uncertainty.
- Entropy=−(pYes​log2​pYes ​+ pNo​log2​pNo​)
1. High Entropy = Mixed data ❌
2. Low Entropy = Pure data ✅

| Entropy | Meaning                        |
| ------- | ------------------------------ |
| 0       | Completely Pure                |
| 0.5     | Partially Mixed                |
| 1       | Maximum Disorder (Binary Case) |

### Information Gain (IG)
- Information Gain tells us: `How much impurity is reduced after a split?`
- IG = Entropy(Parent) − Weighted Entropy(Children)
- A higher Information Gain means a better split.

### Gini Index  
- Gini=1 − ∑p^2
- Another way to measure impurity.

| Feature                 | Entropy   | Gini                   |
| ----------------------- | --------- | ---------------------- |
| Formula                 | Uses log₂ | Uses squares           |
| Speed                   | Slower    | Faster                 |
| Range                   | 0 to 1    | 0 to 0.5 (binary case) |
| Default in scikit-learn | Optional  | Default (historically) |

    Q1. What is Entropy?

    Answer: Entropy measures the impurity or uncertainty in a dataset. Lower entropy means purer data.

        Entropy হলো একটি পরিমাপ, যা বলে ডেটা কতটা মিশ্রিত (Impure) বা অনিশ্চিত (Uncertain)।
        সহজ ভাষায়:
        Entropy = ডেটা কতটা এলোমেলো (Mixed) তা মাপার স্কোর।
        Entropy বেশি = ডেটা বেশি মিশ্রিত
        Entropy কম = ডেটা বেশি পরিষ্কার (Pure)

    Q2. What is Information Gain?

    Answer: Information Gain is the reduction in entropy after a split. The split with the highest Information Gain is preferred.

        যে প্রশ্নটি Entropy সবচেয়ে বেশি কমায়, সেটিই বেছে নেওয়া হয়।
        এই কমার পরিমাণই হলো Information Gain (IG)।
        Information Gain = Split করার পরে কতটা "উন্নতি" হলো।

        Information Gain বেশি ⇒ ভালো Split

    Q3. What is the Gini Index?

    Answer: Gini Index measures impurity using class probabilities. Lower Gini indicates a purer node.

        কম Gini ⇒ ভালো Node

    Q4. Which is faster: Entropy or Gini?

    Answer: Gini is generally faster because it avoids logarithmic calculations.

    Q5. What is the goal of a Decision Tree during splitting?

    Answer: To create child nodes that are as pure as possible by maximizing Information Gain or minimizing impurity.


### Hyperparameter

- criterion → How to choose the best split. It decides how to measure the quality of a split
- max_depth → Maximum tree depth. Controls the maximum depth (height) of the tree.
- min_samples_split → Minimum samples needed to split a node.
- min_samples_leaf → Minimum samples required in each leaf.
- ccp_alpha → Controls pruning.

- 🌳 max_depth → গাছ কত লম্বা হবে।
- ✂️ ccp_alpha → অতিরিক্ত ডাল কেটে ফেলা (pruning)।
- 🍃 min_samples_leaf → প্রতিটি পাতায় (leaf) ন্যূনতম কতটি নমুনা থাকবে।
- 🌿 min_samples_split → নতুন ডাল বের করার জন্য ন্যূনতম কতটি নমুনা দরকার।
- 🔍 criterion → কোন নিয়মে ভালো split নির্বাচন করা হবে (Gini/Entropy)।


- If max_depth is too high:
    Very deep tree,
    Learns noise,
    Overfitting
- If max_depth is too low:
    Very shallow tree,
    Misses important patterns,
    Underfitting,

- min_samples_split `Increase` it:
    Fewer splits.
    Simpler tree.
    Less overfitting
- min_samples_split `Decrease` it:
    More splits,
    More complex tree,
    Higher risk of overfitting

- max_features
What does it do?

Limits how many features are considered when looking for the best split.

#### ccp_alpha (Cost-Complexity Pruning)
What does it do?

Prunes unnecessary branches after the tree is built.

- `Increase` ccp_alpha:
    More pruning
    Simpler tree
    Lower variance
    May underfit if too large
- `Decrease` ccp_alpha:
    Less pruning
    Larger tree
    Higher risk of overfitting

#### How to Prevent Overfitting?
- Stops the tree from growing too deep. DecisionTreeClassifier(max_depth=5)
- Increase min_samples_split
- Increase min_samples_leaf
- Use ccp_alpha (Pruning).Cuts away branches that add little value.

| Hyperparameter      | Increase                       | Decrease                      |
| ------------------- | ------------------------------ | ----------------------------- |
| `max_depth`         | Simpler tree, less overfitting | Deeper tree, more overfitting |
| `min_samples_split` | Fewer splits                   | More splits                   |
| `min_samples_leaf`  | Bigger leaves                  | Smaller leaves                |
| `max_features`      | Faster, more randomness        | Uses more features per split  |
| `max_leaf_nodes`    | Limits tree size               | Allows more leaves            |
| `ccp_alpha`         | More pruning                   | Less pruning                  |

#### Pre-pruning vs Post-pruning

- Pre-pruning (Stop Early)
The tree is prevented from growing too much.

Common hyperparameters:

max_depth
min_samples_split
min_samples_leaf
max_leaf_nodes

Advantages:

Faster training
Smaller trees
Less overfitting

- Post-pruning (Grow, Then Cut)

Grow a large tree.
Remove weak branches.


| Model        | Bias     | Variance |
| ------------ | -------- | -------- |
| Underfitting | High     | Low      |
| Good Fit     | Balanced | Balanced |
| Overfitting  | Low      | High     |

- Bias হলো মডেল কতটা সহজ (simple) হওয়ার কারণে আসল সম্পর্ক শিখতে ব্যর্থ হচ্ছে।

👉 অর্থাৎ, মডেল কম শিখছে (Under Learning)।

- Variance হলো মডেল ট্রেনিং ডেটা অতিরিক্ত মুখস্থ (memorize) করে ফেলে।

👉 অর্থাৎ, নতুন ডেটায় ভালো কাজ করতে পারে না।

Q1. Why do Decision Trees overfit easily?

Answer: Because they can keep splitting until leaves become nearly pure, capturing noise in the training data.

Q2. Which hyperparameter is commonly used for pruning?

Answer: ccp_alpha.

Q3. Name two hyperparameters that help reduce overfitting.

Answer: max_depth and min_samples_leaf (also min_samples_split).

Q4. What is the difference between pre-pruning and post-pruning?

Answer:
- mPre-pruning: Stops tree growth early using constraints.
- Post-pruning: Builds the tree first, then removes unnecessary branches.

Q5. Which is better: a very deep tree or a shallow tree?

Answer: Neither by default. The goal is a balanced tree that generalizes well to unseen data.
