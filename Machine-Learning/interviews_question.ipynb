## Simple Definition

### What is overfitting?
- Model learns noise instead of pattern.

### What is underfitting?
- Model is too simple to learn patterns.

### Bias ex: Underfitting

Bias is the error caused by wrong assumptions in the model. A high bias model is too simple and cannot capture the relationship in data.

Bias হলো model-এর ভুল ধারণার কারণে হওয়া error।

High bias হলে model খুব simple হয় এবং data-এর pattern শিখতে পারে না।

### Variance ex: Overfitting

Variance is the error caused by the model being too sensitive to training data. A high variance model learns noise instead of the real pattern.

Variance হলো training data-এর উপর model-এর অতিরিক্ত নির্ভরতা।

High variance হলে model training data মুখস্থ করে ফেলে এবং নতুন data-তে ভালো কাজ করে না।


`→ High Variance` ## Overfiting

    Train error: Low
    Test error: High

`→ High Bias` ## Underfitting

    Train error: High
    Test error: High

Regularization হলো এমন একটি পদ্ধতি যেখানে Loss Function-এর সাথে অতিরিক্ত penalty যোগ করা হয়, যাতে Model খুব বেশি complex না হয় এবং training data মুখস্থ না করে।

## Decision Tree
- A Decision Tree is a supervised machine learning algorithm that makes decisions by asking a series of questions by recursively splitting data into smaller groups based on feature values. It looks like an upside-down tree.

### Root Node 🌱
1. The first question.
2. Represents the entire dataset.
3. There is only one root node.

### Parent Node 👨‍👧
- A node that splits into other nodes.

### Child Node 👶
- A node created from a parent node.

### Leaf Node 🍃
- The final prediction.

### Branch 🌿
- The connection between nodes.

### Q1. What is a leaf node?
- Answer: A node with no children that contains the final prediction.

### Q2. What is a pure node?
- Answer: A node where all samples belong to the same class.

### Q3. How does a Decision Tree decide where to split?
- Answer: It evaluates many possible splits and chooses the one that best separates the data using measures such as Gini Index or Information Gain.

### Q4. Can a Decision Tree have multiple root nodes?
- Answer: No. A Decision Tree always has exactly one root node.

### 🌳 Decision Tree - Entropy, Information Gain & Gini Index
- This is the heart of Decision Trees.

- Entropy → ডেটা কতটা মিশ্রিত বা অনিশ্চিত তা মাপে।
- Information Gain → Split করার পরে Entropy কতটা কমেছে তা মাপে।
- Gini Index → Entropy-এর মতোই Impurity Measure, তবে এটি দ্রুত গণনা করা যায়।
- Decision Tree-এর লক্ষ্য → এমন Split নির্বাচন করা যাতে Child Nodes যতটা সম্ভব Pure হয়। এই কারণেই Decision Tree Information Gain সর্বোচ্চ বা Gini Impurity সর্বনিম্ন এমন Split বেছে নেয়।

### Entropy
- Entropy measures disorder or uncertainty.
- Entropy=−(pYes​log2​pYes ​+ pNo​log2​pNo​)
1. High Entropy = Mixed data ❌
2. Low Entropy = Pure data ✅

| Entropy | Meaning                        |
| ------- | ------------------------------ |
| 0       | Completely Pure                |
| 0.5     | Partially Mixed                |
| 1       | Maximum Disorder (Binary Case) |

### Information Gain (IG)
- Information Gain tells us: `How much impurity is reduced after a split?`
- IG = Entropy(Parent) − Weighted Entropy(Children)
- A higher Information Gain means a better split.

### Gini Index  
- Gini=1 − ∑p^2
- Another way to measure impurity.

| Feature                 | Entropy   | Gini                   |
| ----------------------- | --------- | ---------------------- |
| Formula                 | Uses log₂ | Uses squares           |
| Speed                   | Slower    | Faster                 |
| Range                   | 0 to 1    | 0 to 0.5 (binary case) |
| Default in scikit-learn | Optional  | Default (historically) |

    Answer: Entropy measures the impurity or uncertainty in a dataset. Lower entropy means purer data.

        Entropy হলো একটি পরিমাপ, যা বলে ডেটা কতটা মিশ্রিত (Impure) বা অনিশ্চিত (Uncertain)।
        সহজ ভাষায়:
        Entropy = ডেটা কতটা এলোমেলো (Mixed) তা মাপার স্কোর।

        Entropy কম = ডেটা বেশি পরিষ্কার (Pure)

    Q2. What is Information Gain?

    Answer: Information Gain is the reduction in entropy after a split. The split with the highest Information Gain is preferred.

        যে প্রশ্নটি Entropy সবচেয়ে বেশি কমায়, সেটিই বেছে নেওয়া হয়।
        এই কমার পরিমাণই হলো Information Gain (IG)।
        Information Gain = Split করার পরে কতটা "উন্নতি" হলো।

    Q3. What is the Gini Index?

    Answer: Gini Index measures impurity using class probabilities. Lower Gini indicates a purer node.

        কম Gini ⇒ ভালো Node

    Q4. Which is faster: Entropy or Gini?

    Q5. What is the goal of a Decision Tree during splitting?

    Answer: To create child nodes that are as pure as possible by maximizing Information Gain or minimizing impurity.

## Random Forest
- Random: Each tree uses random samples of the data and random subsets of features.
- Forest: It is a collection of many Decision Trees.
- Random Forest uses Bagging (Bootstrap Aggregating), where multiple Decision Trees are trained on different bootstrap samples, and their predictions are combined using majority voting (classification) or averaging (regression).

```bash
Random Forest is an ensemble learning algorithm based on bagging. It creates multiple bootstrap samples, trains one Decision Tree on each sample, randomly selects features at every split, and combines predictions using majority voting for classification or averaging for regression. This reduces overfitting and improves generalization.
```

### Bagging = Bootstrap Aggregating

- Bootstrap → Create many random datasets from the original dataset.
- Aggregating → Combine the predictions from all trees.

- Bagging = Bootstrap + Aggregation

### What is Bagging?
- Bagging (Bootstrap Aggregating) trains multiple models on different bootstrap samples and combines their predictions to improve accuracy and reduce variance.

### What is Bootstrap Sampling?
- Answer: Bootstrap sampling is the process of creating multiple datasets by randomly sampling the original dataset with replacement.
- After selecting a row, it is put back into the dataset, so it can be selected again. Therefore, some rows may appear multiple times, while others may not appear at all.

| Randomness               | Purpose                                     |
| ------------------------ | ------------------------------------------- |
| Bootstrap Sampling       | Different training data for each tree       |
| Random Feature Selection | Different features considered at each split |

### What happens if max_features is too high?
- Answer: Trees become more similar, increasing correlation and the risk of overfitting.

### What happens if max_features is too low?
- Answer: Trees become more diverse and overfitting decreases, but each tree may be weaker, which can reduce overall accuracy if set too low.

### Advantages
    High accuracy
    Reduces overfitting
    Handles large datasets
    Works for classification and regression
    Handles nonlinear relationships well
    Gives feature importance
    
### Disadvantages
    Slower than a single Decision Tree
    Uses more memory
    Less interpretable ("black box")
    Training time increases with many trees

### Out-of-Bag (OOB) Error ⭐ (Very Important)

During bootstrap sampling, not every row is selected.

On average:

- About 63.2% of the original samples appear in one bootstrap dataset.
About 36.8% are left out.

Those left-out samples are called Out-of-Bag (OOB) samples.

- Random Forest can use these OOB samples to estimate model performance without creating a separate validation set.

### What is an OOB sample?
- Answer: An Out-of-Bag sample is a training example that was not selected in a particular bootstrap sample. It can be used to estimate model performance without a separate validation dataset.

### Difference Between min_samples_split and min_samples_leaf

| Parameter           | Meaning                                                           |
| ------------------- | ----------------------------------------------------------------- |
| `min_samples_split` | Minimum samples needed **before a node can split**                |
| `min_samples_leaf`  | Minimum samples that **must remain in each leaf after the split** |

### What is the difference between min_samples_split and min_samples_leaf?

- Answer: min_samples_split controls whether a node can split, while min_samples_leaf controls the minimum number of samples that each resulting leaf must contain.

### SVM
- A hyperplane is a decision boundary that separates different classes in a feature space.

- Margin is the distance from the decision boundary to the nearest data points.
- Many separating boundaries may exist, but SVM chooses the one with the maximum margin.
- A larger margin usually makes the model more robust to noise and helps it perform better on unseen data.
- Generalization means performing well on new, unseen examples.
- Maximizing the margin is the fundamental idea behind SVM.

### Hyperplane (The Decision Boundary of SVM) ⭐⭐⭐⭐⭐

- Hyperplane is the foundation of SVM.
- SVM finds the best hyperplane that separates classes with maximum margin.

### Logistic Regression vs SVM Hyperplane
- Logistic Regression : Boundary that minimizes probability error
- SVM : Boundary that maximizes margin

✅ Support Vectors are the closest points to the hyperplane
✅ They determine the position of the decision boundary
✅ Only Support Vectors affect the SVM model
✅ Removing normal points usually does not change the model
✅ Removing Support Vectors can change the boundary
✅ They lie on the margin boundaries wTx+b=±1

### Hard Margin SVM ⭐⭐⭐⭐⭐

- Hard Margin SVM is the original idea of SVM.It works when data can be perfectly separated without any mistakes.
- Hard Margin SVM finds a hyperplane that perfectly separates two classes with zero classification errors while maximizing the margin.
- Hard Margin SVM finds a maximum-margin hyperplane that perfectly separates classes without any misclassification.

Hard Margin:
- Perfect Separation + Maximum Margin + Zero Error

### What is Slack Variable?
- Slack variable (ξ) measures the amount of margin violation or misclassification allowed for a data point.

### What does C control in SVM?
- C controls the tradeoff between maximizing the margin and minimizing classification errors.

### What happens when C increases?
- The model penalizes errors more, creates a smaller margin, and may overfit.

### What happens when C decreases?
- The model allows more errors, creates a larger margin, and usually generalizes better.

### If C is extremely high, what happens?

- The model tries to classify almost every training sample correctly, producing a narrow margin and increasing the risk of overfitting.

- Large Gamma: Complex boundary -> Overfitting possible(gamma=100)
