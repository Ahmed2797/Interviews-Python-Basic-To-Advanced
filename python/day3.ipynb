### What is a Set?

`A set is an unordered, mutable collection of unique elements in Python.`


1️⃣ What is a Set?

### Common Set Methods

| Method       | Description                       | Example         |
| ------------ | --------------------------------- | --------------- |
| `add(x)`     | Add element                       | `s.add(5)`      |
| `remove(x)`  | Remove element (error if missing) | `s.remove(5)`   |
| `discard(x)` | Remove element (safe if missing)  | `s.discard(5)`  |
| `pop()`      | Remove arbitrary element          | `s.pop()`       |
| `clear()`    | Remove all elements               | `s.clear()`     |
| `copy()`     | Copy set                          | `s2 = s.copy()` |
| `len(s)`     | Count elements                    | `len(s)`        |


### differents between `remove() and pop() and discard()` methold

0. `discard(x)` | Remove element (safe if missing)

1. `remove(x)`  | Remove element (error if missing) 

2. `pop()`      | Remove arbitrary element | return the pop element


```python
a = {1,2,3}
b = {3,4,5}

# Union
print(a | b)  # {1,2,3,4,5}

# Intersection
print(a & b)  # {3}

# Difference
print(a - b)  # {1,2}

# Symmetric Difference
print(a ^ b)  # {1,2,4,5}
```

#### 🔹 Set() vs Frozenset()

`Think of Set as a playground where kids can run around and change things, and Frozenset as a museum exhibit where nothing can move.`

| Feature                     | `set()`                                      | `frozenset()`                             |
| --------------------------- | -------------------------------------------- | ----------------------------------------- |
| **Mutable**                 | ✅ You can add/remove elements                | ❌ Immutable, cannot change                |
| **Hashable**                | ❌ Not hashable                               | ✅ Hashable, can be used as dictionary key |
| **Methods like add/remove** | ✅ `.add()`, `.remove()`, `.discard()`        | ❌ None of these, read-only                |
| **Membership test**         | ✅ `x in s` works                             | ✅ `x in fs` works                         |
| **Use case**                | When you need to change elements dynamically | When you need a fixed set (like dict key) |


```python
## Immutable, hashable → can be dictionary keys

fs = frozenset([1,2,3,3])
print(len(fs)) #0,1,2,3
fs
```

```python
## 🔥 Problem 1: Remove duplicates (Classic)
nums = [1,2,3,2,4,3,5]
unique = list(set(nums))
print(unique)


## 🧠 maintain করতে পারবো?
unique1 = list(dict.fromkeys(nums))
unique1


## 🔥 Problem 3: First Duplicate Element
nums = [1,2,3,4,2,5]
seen = set()

for n in nums:
    if n in seen:
        print(n)
        break
    seen.add(n)
print(seen)


## 🔥 Problem: Missing Number
nums = [1,2,3,5]
n = 5

missing = set(range(1, n+1)) - set(nums)
print(missing.pop())


## 🔥 Problem: Unique Characters in String
def is_unique(s):
    return len(s) == len(set(s))

print(is_unique("abc"))
print(is_unique("aba"))


## 🔥 Problem: Remove Common Words
s1 = "i love python"
s2 = "i love ai"

result = set(s1.split()) - set(s2.split())
print(result)


## Count unique words in a sentence
sentence = "I love Python and I love python"

words = sentence.lower().split()
unique_words = set(words)

print(len(unique_words))


## Find first repeating element
sentence = "I love Python and I love python"
element = [4,3,1,2,3,3,1,3]
## element = sentence.lower().split()

seen = set()
for i in element:
    if i in seen:
        print("found",i)
        break
    seen.add(i)
print(seen)


## Find elements present in only ONE list
A = [1,2,3,4]
B = [3,4,5,6]

result = set(A) ^ set(B)
print(result)


## Common Elements in N Lists
lists = [
    [1,2,3,4],
    [2,3,5],
    [2,3,6]
]
common = set(lists[0])

for lst in lists[1:]:
    common &= set(lst)
print(common)


## Distinct Elements in Window Size K  (SLIDING WINDOW + HASHMAP)
from collections import defaultdict

def count_distinct_in_windows(arr, k):
    """
    Count distinct elements in each sliding window of size k.

    Parameters:
    arr (list): List of integers
    k (int): Size of the sliding window

    Returns:
    dict: Frequency map of the last window
    """

    freq = defaultdict(int)

    # Step 1: First window
    for i in range(k):
        freq[arr[i]] += 1

    # Step 2: Slide the window
    for i in range(k, len(arr)):
        # add new element (right side)
        freq[arr[i]] += 1

        # remove old element (left side)
        freq[arr[i - k]] -= 1

        if freq[arr[i - k]] == 0:
            del freq[arr[i - k]]

    return freq

arr = [1, 2, 4, 4, 5, 6, 6, 7, 8]

k = 5
freq_distinct = count_distinct_in_windows(arr,k)
freq_distinct
```

### Hashability

পাইথনে যে জিনিসের মান কখনো পরিবর্তন করা যায় না (Immutable), সেটিরই একটি স্থায়ী টোকেন নম্বর বা হ্যাশ তৈরি করা সম্ভব, আর তাকেই Hashable বলে।

পাইথনে Hashable (হ্যাশযোগ্য) অবজেক্ট বলতে এমন একটি অবজেক্টকে বোঝায়, যার একটি নির্দিষ্ট হ্যাশ ভ্যালু (Hash Value) থাকে যা তার পুরো জীবনচক্রে (Lifetime) কখনো পরিবর্তন হয় না।

    Hashable = Has an Unchangeable ID.

    Unhashable = Its ID can change.

1. `Hashable` (The Lock)

Some things in Python can never change after you create them (like a Number, a Word, or a Tuple). Because they never change, Python can give them a permanent, fixed ID number (a hash).

    Rule: If it cannot change, it is Hashable.

2. `Unhashable` (The Mud)

Some things in Python can change anytime (like a List or a Dictionary, where you can add or remove items). Because the contents can change, Python cannot give them a permanent ID.

    Rule: If it can change, it is Unhashable.

`Why list cannot be inside a set?`

✅ Reason

1.Set elements must be hashable

2.List is mutable → not hashable

3.Tuple is immutable → hashable


| Feature     | List | Tuple | Set |
| ----------- | ---- | ----- | --- |
| Ordered     | ✅    | ✅     | ❌   |
| Mutable     | ✅    | ❌     | ✅   |
| Duplicate   | ✅    | ✅     | ❌   |
| Index       | ✅    | ✅     | ❌   |
| Fast search | ❌    | ❌     | ✅   |


```python
# ১. টেক্সট বা স্ট্রিং কখনো বদলায় না (Immutable)
print(hash("রনি"))  # আউটপুট: একটি নির্দিষ্ট নম্বর (যেমন: 483920432)
# তারমানে "রনি" একটি Hashable জিনিস।

# ২. লিস্টের ভেতরের জিনিস যেকোনো সময় বদলানো যায় (Mutable)
আমার_লিস্ট = [1, 2, 3]
print(hash(আমার_লিস্ট))  # আউটপুট: TypeError: unhashable type: 'list'
# পাইথন বলবে: "ভাই, এটা তো যেকোনো সময় বদলে যেতে পারে, এর কোনো স্থায়ী টোকেন আমি বানাতে পারবো না!"
```
