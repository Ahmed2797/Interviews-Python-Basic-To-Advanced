### `Instance Variables` and `Class Variables`

#### Difference between class variable and instance variable?
| Class Variable   | Instance Variable      |
| ---------------- | ---------------------- |
| Shared           | Separate               |
| Defined in class | Defined in constructor |

| Class Variable | Instance Variable |
| -------------- | ----------------- |
| Shared         | Personal          |
| Same for all   | Different         |



````python
## Instance Variables
class Car:

    def __init__(self, brand):
        self.brand = brand

c1 = Car("BMW")
c2 = Car("Tesla")

print(c1.brand)
print(c2.brand)
````

````python
## Class Variables

class Student:

    school = "ABC School"

    def __init__(self, name):
        self.name = name

s1 = Student("Rahim")
s2 = Student("Karim")

print(s1.school)
print(s2.school)
````

### `Encapsulation`

- Wrapping data and methods together.
- Use private variables.
- Hide data using private variable.

| Type      | Syntax |
| --------- | ------ |
| Public    | name   |
| Protected | _name  |
| Private   | __name |


````python
class Bank:

    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        self.__balance += amount

    def show(self):
        print(self.__balance)

b = Bank()

b.deposit(500)
b.show()
````

### `Abstraction`

- Hiding internal implementation.

- Use abstract classes.

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        print('Car is ready')

class Car(Vehicle):

    def start(self):
        print("Car Start")

c = Car()
c.start()
````

#### Difference between abstraction and encapsulation?
| Abstraction         | Encapsulation |
| ------------------- | ------------- |
| Hide implementation | Hide data     |


### `Inheritance`

- Child class gets properties from parent class.

````python
class Father:

    def house(self):
        print("Father House")

class Son(Father):
    pass

s = Son()
s.house()
````

````python
class A:
    def show1(self):
        print("A")

class B:
    def show2(self):
        print("B")

class C(A, B):
    pass

c = C()

c.show1()
c.show2()
````

### `super()`

- super() calls parent class methods or constructor.

Why use super()?
- Access parent constructor
- Access parent methods

````python
class Father:

    def __init__(self):
        print("Father Constructor")

class Son(Father):

    def __init__(self):
        super().__init__()
        print("Son Constructor")

s = Son()
````

### `Method Overriding`

- Child changes parent method.

````python
class Animal:

    def sound(self):
        print("Animal Sound")

class Dog(Animal):

    def sound(self):
        print("Bark")

d = Dog()
d.sound()
````

### Difference between method overloading and overriding?
| Overloading             | Overriding                  |
| ----------------------- | --------------------------- |
| Same method diff params | Child changes parent method |


### `Method Overloading`



````python
class Math:

    def add(self, a, b, c=0):
        return a+b+c

m = Math()

print(m.add(1,2))
print(m.add(1,2,3))
````

### `Polymorphism`

- One thing many forms.
- Same method name, different behavior.

````python
class Bird:

    def fly(self):
        print("Bird Fly")

class Airplane:

    def fly(self):
        print("Airplane Fly")

for obj in [Bird(), Airplane()]:
    obj.fly()
````

````python
class Cat:

    def sound(self):
        print("Meow")

class Dog:

    def sound(self):
        print("Bark")

for animal in [Cat(), Dog()]:
    animal.sound()
````

### `Magic / Dunder Methods`

| Method     | Meaning                 |
| ---------- | ----------------------- |
| `__init__` | Constructor             |
| `__str__`  | String display          |
| `__len__`  | Length                  |
| `__add__`  | + operator              |
| `__repr__` | Official representation |
| `__del__`  | Destructor              |
| `__call__` | Callable object         |
| Method     | Uses                     |
| ---------- | ------------------------ |
| `__init__` | Constructor              |
| `__str__`  | Object print             |
| `__repr__` | Developer representation |
| `__len__`  | Length                   |
| `__add__`  | + operator               |

````python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

s = Student("Tanvir")

print(s)
````

````python
class Team:

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

    # def lenth(self):
    #     return len(self.members)

t = Team(["a", "b", "c"])

print(len(t))
# t.lenth()
````

````python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
````

````python
class Student:

    def __init__(self):
        self.__age = 0

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        self.__age = value

s = Student()

s.age = 20

print(s.age)
````

### `isinstance() and issubclass()`

- isinstance(object, class) checks if an object belongs to a specific class.
- issubclass(child, parent) checks if the first class is a subclass of the second class.

````python
class A:
    pass

class B(A):
    pass

b = B()

print(isinstance(b, B))
print(isinstance(b, A))

print(issubclass(B, A))
````

### `MRO (Method Resolution Order)` Method searching order in inheritance.
- Python searches methods in a specific order.

````python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(D.mro())
````

### `Duck Typing`

- Object type doesn't matter.
- Behavior matters.

````python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

for animal in [Dog(), Cat()]:
    animal.sound()
````

````python
class Employee:

    company = "Google"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show(self):
        print(self.name, self.salary,'super(). call show() methold')

    @classmethod
    def company_name(cls):
        print(cls.company)

    @staticmethod
    def greet():
        print("Welcome")

class Developer(Employee):

    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language

    def show(self):
        super().show()
        print(self.language)

d = Developer("Tanvir", 5000, "Python")

d.show()
````

### `Generator`

Function using `yield`. 
- yield = pause and continue
- Memory efficient.

### yield vs return

| yield            | return            |
| ---------------- | ----------------- |
| One by one       | Full return       |
| Memory efficient | Stores everything |


````python
def test():

    yield 1
    yield 2
    yield 3

g = test()

print(next(g))
print(next(g))
````

### `Decorators`

What is Decorator?
- Function modifying another function.

````python
def decorator(func):

    def wrapper():
        print("Before")
        func()
        print("After")

    return wrapper

@decorator
def hello():
    print("Hello")

hello()
````

### `Factory Pattern`

- Creates objects without exposing creation logic.

````python
class Dog:
    pass

class Cat:
    pass

class Factory:

    def get_animal(self, name):

        if name == "dog":
            return Dog()

        return Cat()
````
